```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Make charts look clean and professional
sns.set_style("darkgrid")
plt.rcParams['figure.figsize'] = (10, 6)

print("Libraries ready!")
```

```python
df = pd.read_csv('../data/creditcard.csv')
print(f"Dataset loaded! Shape: {df.shape}")
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='Class', data=df, hue='Class', palette=['#2ecc71', '#e74c3c'], legend=False)
plt.title('Fraud vs Normal Transactions', fontsize=16, fontweight='bold')
plt.xticks([0, 1], ['Normal (0)', 'Fraud (1)'], fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.show()

print(f"Normal: {df['Class'].value_counts()[0]:,}")
print(f"Fraud:  {df['Class'].value_counts()[1]:,}")
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Normal transactions
ax1.hist(df[df['Class']==0]['Amount'], bins=50, color='#2ecc71', edgecolor='black')
ax1.set_title('Normal Transaction Amounts', fontsize=14, fontweight='bold')
ax1.set_xlabel('Amount')
ax1.set_ylabel('Count')

# Fraud transactions
ax2.hist(df[df['Class']==1]['Amount'], bins=50, color='#e74c3c', edgecolor='black')
ax2.set_title('Fraud Transaction Amounts', fontsize=14, fontweight='bold')
ax2.set_xlabel('Amount')
ax2.set_ylabel('Count')

plt.tight_layout()
plt.show()

print(f"Average Normal Amount: ₹{df[df['Class']==0]['Amount'].mean():.2f}")
print(f"Average Fraud Amount:  ₹{df[df['Class']==1]['Amount'].mean():.2f}")
```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 8))

# Normal transactions over time
ax1.plot(df[df['Class']==0]['Time'], 
         df[df['Class']==0]['Amount'], 
         '.', color='#2ecc71', alpha=0.3, markersize=2)
ax1.set_title('Normal Transactions Over Time', fontsize=14, fontweight='bold')
ax1.set_xlabel('Time')
ax1.set_ylabel('Amount')

# Fraud transactions over time
ax2.plot(df[df['Class']==1]['Time'], 
         df[df['Class']==1]['Amount'], 
         '.', color='#e74c3c', alpha=0.5, markersize=3)
ax2.set_title('Fraud Transactions Over Time', fontsize=14, fontweight='bold')
ax2.set_xlabel('Time')
ax2.set_ylabel('Amount')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 10))
correlation = df.corr()
sns.heatmap(correlation, 
            cmap='coolwarm',
            center=0,
            linewidths=0.5)
plt.title('Correlation Heatmap of All Features', fontsize=16, fontweight='bold')
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Class', y='Amount', data=df, 
            hue='Class',
            palette=['#2ecc71', '#e74c3c'],
            legend=False)
plt.title('Transaction Amount Distribution: Normal vs Fraud', fontsize=16, fontweight='bold')
plt.xticks([0, 1], ['Normal', 'Fraud'], fontsize=12)
plt.ylabel('Amount (₹)', fontsize=12)
plt.show()

print("Key Statistics:")
print(f"\nNormal Transactions:")
print(f"  Max Amount:    ₹{df[df['Class']==0]['Amount'].max():.2f}")
print(f"  Min Amount:    ₹{df[df['Class']==0]['Amount'].min():.2f}")
print(f"  Avg Amount:    ₹{df[df['Class']==0]['Amount'].mean():.2f}")
print(f"\nFraud Transactions:")
print(f"  Max Amount:    ₹{df[df['Class']==1]['Amount'].max():.2f}")
print(f"  Min Amount:    ₹{df[df['Class']==1]['Amount'].min():.2f}")
print(f"  Avg Amount:    ₹{df[df['Class']==1]['Amount'].mean():.2f}")
```
